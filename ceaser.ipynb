```python
def encrypt(text, shift):
    result = ""

    for char in text:
        if char.isalpha():
            # Check if the character is uppercase or lowercase
            base = ord('A') if char.isupper() else ord('a')

            # Shift the character
            new_char = chr((ord(char) - base + shift) % 26 + base)

            result += new_char
        else:
            # Keep spaces, numbers, and symbols unchanged
            result += char

    return result


def deccrypt(cipher, shift):
    result = ""

    for char in cipher:
        if char.isalpha():
            # Check if the character is uppercase or lowercase
            base = ord('A') if char.isupper() else ord('a')

            # Shift the character
            new_char = chr((ord(char) - base - shift) % 26 + base)

            result += new_char
        else:
            # Keep spaces, numbers, and symbols unchanged
            result += char

    return result




# User Input
text = input("Enter the text to encrypt: ")
shift = int(input("Enter the shift value: "))

cipher = encrypt(text, shift)
plain=deccrypt(cipher,shift)

print("\nEncrypted Text:", cipher,"\nPlain Text:",plain)
```
